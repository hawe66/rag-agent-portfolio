# PDF Loader Comparison

**Goal:** Compare PyPDFLoader vs PDFPlumberLoader for Korean text extraction quality.

**Test File:** `waterpurifier_complex.pdf` (1.7 MB)  
**Pages:** 9-12 (actual content)

## 1. Setup

````python
# Install dependencies if needed
# !pip install pypdf pdfplumber langchain langchain-community
````

````python
import time
from pathlib import Path

PDF_PATH = Path("../data/raw_pdfs/waterpurifier_complex.pdf")
TARGET_PAGES = [9, 10, 11, 12]  # 1-indexed for display, 0-indexed internally

print(f"PDF exists: {PDF_PATH.exists()}")
print(f"PDF size: {PDF_PATH.stat().st_size / 1024 / 1024:.2f} MB")
````

## 2. PyPDFLoader

````python
from langchain_community.document_loaders import PyPDFLoader

start = time.time()
pypdf_loader = PyPDFLoader(str(PDF_PATH))
pypdf_docs = pypdf_loader.load()
pypdf_time = time.time() - start

print(f"PyPDFLoader: {len(pypdf_docs)} pages loaded in {pypdf_time:.2f}s")
````

````python
# Extract target pages (0-indexed)
pypdf_target = [pypdf_docs[i-1] for i in TARGET_PAGES if i-1 < len(pypdf_docs)]

for doc in pypdf_target:
    page_num = doc.metadata.get('page', 'unknown') + 1  # Convert to 1-indexed
    print(f"\n{'='*60}")
    print(f"PAGE {page_num} (PyPDFLoader)")
    print(f"{'='*60}")
    print(doc.page_content[:2000])  # First 2000 chars
    print(f"\n... [Total: {len(doc.page_content)} chars]")
````

## 3. PDFPlumberLoader

````python
from langchain_community.document_loaders import PDFPlumberLoader

start = time.time()
plumber_loader = PDFPlumberLoader(str(PDF_PATH))
plumber_docs = plumber_loader.load()
plumber_time = time.time() - start

print(f"PDFPlumberLoader: {len(plumber_docs)} pages loaded in {plumber_time:.2f}s")
````

````python
# Extract target pages (0-indexed)
plumber_target = [plumber_docs[i-1] for i in TARGET_PAGES if i-1 < len(plumber_docs)]

for doc in plumber_target:
    page_num = doc.metadata.get('page', 'unknown') + 1  # Convert to 1-indexed
    print(f"\n{'='*60}")
    print(f"PAGE {page_num} (PDFPlumberLoader)")
    print(f"{'='*60}")
    print(doc.page_content[:2000])  # First 2000 chars
    print(f"\n... [Total: {len(doc.page_content)} chars]")
````

## 4. Side-by-Side Comparison

````python
# Compare character counts and content differences
print("Character Count Comparison:\n")
print(f"{'Page':<8} {'PyPDF':<12} {'Plumber':<12} {'Diff':<12}")
print("-" * 44)

for i, page in enumerate(TARGET_PAGES):
    pypdf_len = len(pypdf_target[i].page_content) if i < len(pypdf_target) else 0
    plumber_len = len(plumber_target[i].page_content) if i < len(plumber_target) else 0
    diff = plumber_len - pypdf_len
    print(f"{page:<8} {pypdf_len:<12} {plumber_len:<12} {diff:+d}")
````

````python
# Performance comparison
print("\nPerformance Comparison:\n")
print(f"PyPDFLoader:     {pypdf_time:.2f}s")
print(f"PDFPlumberLoader: {plumber_time:.2f}s")
````

## 5. Image Detection (pdfplumber)

````python
import pdfplumber

with pdfplumber.open(PDF_PATH) as pdf:
    print(f"Total pages: {len(pdf.pages)}\n")
    
    for page_num in TARGET_PAGES:
        page = pdf.pages[page_num - 1]  # 0-indexed
        images = page.images
        
        print(f"Page {page_num}: {len(images)} image(s)")
        
        for idx, img in enumerate(images):
            # Bounding box: x0, top, x1, bottom
            bbox = (img['x0'], img['top'], img['x1'], img['bottom'])
            width = img['x1'] - img['x0']
            height = img['bottom'] - img['top']
            print(f"  [{idx}] bbox={bbox}, size=({width:.1f} x {height:.1f})")
````

### Known Limitations (to revisit later)

**LIM-001: Text Order Mismatch (SEVERE)**

Both loaders fail on multi-column layouts. Text from different columns gets interleaved, **breaking sentences mid-way**.

Example from Page 11 (two-column: left=steps, right=table):

```
# Extracted (wrong):
2
제품의 3 m 거리 이내에서 'Hi LG'(하이 엘지)를
기능 명령어 (예시)        ← table header injected mid-sentence
말하세요.                 ← sentence continuation, now orphaned
오늘/일주일간/이번 달의 물 사용량을
• 작동음이 울리고 음성 명령을 들을 준비 상태가 되면 확인할 수 있습니다.
물 사용량                 ← table cell mixed in
제어창의 x 아이콘이 깜빡입니다. 조회 • 오늘 물 사용량 알려줘
```

```
# Expected (correct):
[Left column - Steps]
2. 제품의 3 m 거리 이내에서 'Hi LG'(하이 엘지)를 말하세요.
   • 작동음이 울리고 음성 명령을 들을 준비 상태가 되면 제어창의 x 아이콘이 깜빡입니다.

[Right column - Table]
| 기능 | 명령어 (예시) |
| 물 사용량 조회 | 오늘 물 사용량 알려줘 / 이번 주 냉수 사용량 알려줘 |
```

**Impact:**
- Chunking will produce incoherent text blocks
- Retrieval quality will degrade (semantic search on broken sentences)
- LLM generation will hallucinate or fail to answer correctly

**Potential solutions:**
- Layout-aware parsing: `unstructured`, `docling`, `marker-pdf`
- Vision LLM-based extraction (treat PDF page as image)
- Manual post-processing with heuristics (fragile)

---

**LIM-002: Incomplete Image Detection**
- Both pdfplumber and PyMuPDF miss some images
- Likely causes: vector graphics, nested XObjects, or non-standard embeddings
- Impact: Phase 2 cross-modal alignment may miss image-text pairs
- Potential solutions: pdf2image (rasterize full page), vision LLM to identify image regions

### Decision

**Selected Loader:** PDFPlumberLoader

**Reasoning:**
1. Slightly faster (~20%)
2. Korean text extraction quality is equivalent to PyPDFLoader
3. Built-in image/table detection APIs (useful for future exploration)
4. Text order issue affects both loaders equally — not a differentiator
